```python
def cost(board):
    c = 0
    for i in range(4):
        for j in range(i + 1, 4):
            if board[i] == board[j] or abs(board[i] - board[j]) == abs(i - j):
                c += 1
    return c

def hill_climbing():
    board = [1, 3, 0, 2]

    while True:
        current = cost(board)

        if current == 0:
            print("Solution:", board)
            return

        best = board
        best_cost = current

        for col in range(4):
            for row in range(4):
                new = board.copy()
                new[col] = row
                c = cost(new)

                if c < best_cost:
                    best = new
                    best_cost = c

        if best_cost == current:
            print("Local minimum reached")
            return

        board = best

hill_climbing()
```
