```python

# Forward Reasoning (Forward Chaining)

# Step 1: Initialize the facts
facts = {"It is raining"}

# Step 2: Define production rules
rules = [
    ({"It is raining"}, "Roads are wet"),
    ({"Roads are wet"}, "Traffic is slow"),
    ({"Traffic is slow"}, "People are late"),
    ({"People are late"}, "Meeting is delayed")
]

# Step 3: Define the goal
goal = "Meeting is delayed"

# Step 4: Forward Chaining Algorithm
def forward_chaining(facts, rules, goal):

    print("Initial Facts:", facts)

    while True:
        new_fact_added = False

        for condition, conclusion in rules:

            # Check whether rule conditions are satisfied
            if condition.issubset(facts):

                # Check whether conclusion is already present
                if conclusion not in facts:

                    # Add new fact
                    facts.add(conclusion)

                    print("New Fact Generated:", conclusion)

                    new_fact_added = True

                    # Check whether goal is achieved
                    if goal in facts:
                        print("\nGoal Achieved!")
                        return True

        # Stop if no new facts are generated
        if not new_fact_added:
            break

    print("\nGoal Cannot Be Achieved!")
    return False


# Step 5: Execute the algorithm
result = forward_chaining(facts, rules, goal)

print("\nFinal Facts:", facts)
print("Result:", result)
```
