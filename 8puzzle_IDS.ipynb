```python
def ids(start, goal):
    def depth_limited_search(state, depth, path, visited):
        if state == goal:
            return path

        if depth == 0:
            return None

        visited.add(state)

        zero = state.index(0)
        row, col = divmod(zero, 3)

        moves = [
            (-1, 0),  # Up
            (1, 0),   # Down
            (0, -1),  # Left
            (0, 1)    # Right
        ]

        for dr, dc in moves:
            nr, nc = row + dr, col + dc

            if 0 <= nr < 3 and 0 <= nc < 3:
                new_zero = nr * 3 + nc

                new_state = list(state)
                new_state[zero], new_state[new_zero] = \
                    new_state[new_zero], new_state[zero]

                new_state = tuple(new_state)

                if new_state not in visited:
                    result = depth_limited_search(
                        new_state,
                        depth - 1,
                        path + [new_state],
                        visited
                    )

                    if result:
                        return result

        return None

    depth = 0

    while True:
        result = depth_limited_search(
            start,
            depth,
            [start],
            set()
        )

        if result:
            return result

        depth += 1


# 0 represents the blank space
start = (
    1, 2, 3,
    4, 0, 6,
    7, 5, 8
)

goal = (
    1, 2, 3,
    4, 5, 6,
    7, 8, 0
)

solution = ids(start, goal)

print("Solution:")

for state in solution:
    print(state[0:3])
    print(state[3:6])
    print(state[6:9])
    print()
```
