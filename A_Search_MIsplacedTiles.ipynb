```python
import heapq

def misplaced_tiles(state, goal):
    return sum(1 for i in range(9)
               if state[i] != 0 and state[i] != goal[i])


def get_neighbors(state):
    neighbors = []
    zero = state.index(0)
    row, col = zero // 3, zero % 3

    moves = [(-1, 0), (1, 0), (0, -1), (0, 1)]

    for dr, dc in moves:
        r, c = row + dr, col + dc

        if 0 <= r < 3 and 0 <= c < 3:
            new_zero = r * 3 + c
            new_state = list(state)
            new_state[zero], new_state[new_zero] = (
                new_state[new_zero], new_state[zero]
            )
            neighbors.append(tuple(new_state))

    return neighbors


def display(state):
    for i in range(0, 9, 3):
        print("  " + " ".join(
            "_" if x == 0 else str(x) for x in state[i:i + 3]
        ))


def a_star(start, goal):
    pq = [(misplaced_tiles(start, goal), 0, start)]
    parent = {start: None}
    g_cost = {start: 0}

    while pq:
        f, g, current = heapq.heappop(pq)

        if g > g_cost[current]:
            continue

        if current == goal:
            path = []

            while current is not None:
                path.append(current)
                current = parent[current]

            return path[::-1]

        for neighbor in get_neighbors(current):
            new_g = g + 1

            if neighbor not in g_cost or new_g < g_cost[neighbor]:
                g_cost[neighbor] = new_g
                parent[neighbor] = current

                h = misplaced_tiles(neighbor, goal)
                new_f = new_g + h

                heapq.heappush(pq, (new_f, new_g, neighbor))

    return None


start = (
    1, 2, 3,
    4, 0, 6,
    7, 5, 8
)

goal = (
    1, 2, 3,
    4, 5, 6,
    7, 8, 0
)

print("=" * 45)
print("   8-PUZZLE USING A* ALGORITHM")
print("   HEURISTIC: MISPLACED TILES")
print("=" * 45)

print("\nINITIAL STATE:")
display(start)

print("\nGOAL STATE:")
display(goal)

print("\nINITIAL MISPLACED TILES:",
      misplaced_tiles(start, goal))

path = a_star(start, goal)

if path:
    print("\n" + "=" * 45)
    print("       STEP-BY-STEP SOLUTION")
    print("=" * 45)

    for step, state in enumerate(path):
        g = step
        h = misplaced_tiles(state, goal)
        f = g + h

        print(f"\nSTEP {step}")
        print("-" * 30)
        display(state)

        print(f"g(n) = {g}  (Cost from start)")
        print(f"h(n) = {h}  (Misplaced tiles)")
        print(f"f(n) = {f}  (g(n) + h(n))")

    print("\n" + "=" * 45)
    print("SOLUTION FOUND!")
    print("Total Moves:", len(path) - 1)
    print("=" * 45)

else:
    print("No solution found.")
```
