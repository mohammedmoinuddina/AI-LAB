```python
room_A = "Dirty"
room_B = "Dirty"
vacuum = "A"

print("Initial State:")
print("Room A:", room_A)
print("Room B:", room_B)
print("Vacuum:", vacuum)

# Clean Room A
if vacuum == "A" and room_A == "Dirty":
    print("\nSuck")
    room_A = "Clean"

# Move to Room B
if vacuum == "A":
    print("Move Right")
    vacuum = "B"

# Clean Room B
if vacuum == "B" and room_B == "Dirty":
    print("Suck")
    room_B = "Clean"

print("\nFinal State:")
print("Room A:", room_A)
print("Room B:", room_B)
print("Vacuum:", vacuum)
```
